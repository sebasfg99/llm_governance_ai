# POST /ai/query — Payloads de prueba

**Endpoint:** `POST /ai/query`  
**Auth:** Bearer token en header `Authorization`  
**Content-Type:** `application/json`

---

## Campos del request

| Campo | Tipo | Default | Notas |
|---|---|---|---|
| `query` | string | requerido | 1–4000 chars |
| `use_case` | string | `chatbot` | chatbot, executive_summary, policy_generation, risk_intake, risk_analysis, controls_generation |
| `top_k` | int | `5` | 1–20 |
| `include_global` | bool | `true` | incluir biblioteca global |
| `response_mode` | string | `text` | `text` o `structured` |
| `country` | string | null | slug: `costa-rica`, `panama`, etc. |
| `project_id` | UUID | null | memoria L2 (expediente) |
| `thread_id` | UUID | null | memoria L1 (historial de conversación) |
| `framework_filter` | list[string] | null | `iso_42001`, `nist_airmf`, `owasp`, etc. |
| `stream` | bool | `false` | SSE token-by-token, incompatible con `structured` |
